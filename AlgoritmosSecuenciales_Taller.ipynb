## 3.3. Dados 3 puntos cualesquiera en un plano, puntos={`x1, y1, x2, y2, x3, y3`} los cuales forman un triángulo. Hacer un programa para calcular el área del triángulo (sugerencia, use la fórmula de Héron).

```python
import math
import matplotlib.pyplot as plt

def distancia(x1, y1, x2, y2):
    '''
    La distancia entre dos puntos : p1(x1,y1) y p2(x2,y2) se calcula con la formula: 
    D=RAIZ((X2-X1)^2 + (Y2-Y1)^2)
    '''
    return math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

def area_triangulo_heron(x1, y1, x2, y2, x3, y3):
    # Calcular longitudes de los lados
    a = distancia(x1, y1, x2, y2) #P1 -> P2
    b = distancia(x2, y2, x3, y3) #P2 -> P3
    c = distancia(x3, y3, x1, y1) #P3 -> P1

    # Calcular el semiperímetro del triangulo
    s = (a + b + c) / 2

    # Aplicar la fórmula de Héron
    area = math.sqrt(s * (s - a) * (s - b) * (s - c))
    # Mostrar el resultado
    print(f"El área del triángulo es: {area:.4f} U2")
    return area

def graficar_triangulo(x1, y1, x2, y2, x3, y3):
    plt.figure(figsize=(6,6))  # Tamaño de la gráfica
    
    # Coordenadas de los puntos
    x = [x1, x2, x3, x1]  # Cierra el triángulo regresando al punto inicial
    y = [y1, y2, y3, y1]

    # Dibujar el triángulo
    plt.plot(x, y, marker='o', linestyle='-', color='black', markersize=8, label="Triángulo")
    
    # Etiquetas de los vértices
    plt.text(x1, y1, ' P1', fontsize=12, color='purple', fontweight='bold')
    plt.text(x2, y2, ' P2', fontsize=12, color='purple', fontweight='bold')
    plt.text(x3, y3, ' P3', fontsize=12, color='purple', fontweight='bold')

    # Ajustes del gráfico
    plt.xlabel("Eje X")
    plt.ylabel("Eje Y")
    plt.title("Triángulo formado")
    plt.grid(True)
    plt.legend()
    plt.show()
```

```python
# Solicitar los puntos al usuario
x1, y1 = map(float, input("Ingrese las coordenadas de P1 (x1 y1): ").split())
x2, y2 = map(float, input("Ingrese las coordenadas de P2 (x2 y2): ").split())
x3, y3 = map(float, input("Ingrese las coordenadas de P3 (x3 y3): ").split())
area = area_triangulo_heron(x1, y1, x2, y2, x3, y3)
```

```python
def convertir_tiempo(segundos):
    horas = segundos // 3600  # Obtener horas
    segundos_restantes = segundos % 3600  # Segundos restantes después de extraer las horas
    minutos = segundos_restantes // 60  # Obtener minutos
    segundos_finales = segundos_restantes % 60  # Obtener los segundos restantes
    print(f"{segundos} segundos equivalen a: {horas} hora(s), {minutos} minuto(s) y {segundos_finales} segundo(s).")

    return horas, minutos, segundos_finales
```

## 3.5. De la ecuación cuadrática obtener las raíces, el vértice

```python
import math
import numpy as np
import matplotlib.pyplot as plt

# Función para calcular las raíces y el vértice de una ecuación cuadrática
def resolver_ecuacion_cuadratica(a, b, c):
    # Calculamos el discriminante
    discriminante = b**2 - 4*a*c

    # Calculamos el vértice
    x_vertice = -b / (2*a)
    y_vertice = a * x_vertice**2 + b * x_vertice + c

    # Verificamos el discriminante para encontrar las raíces
    if discriminante > 0:
        raiz1 = (-b + math.sqrt(discriminante)) / (2*a)
        raiz2 = (-b - math.sqrt(discriminante)) / (2*a)
        tipo_raices = "Dos raíces reales y diferentes"
    elif discriminante == 0:
        raiz1 = raiz2 = -b / (2*a)
        tipo_raices = "Una raíz real doble"
    else:
        parte_real = -b / (2*a)
        parte_imaginaria = math.sqrt(abs(discriminante)) / (2*a)
        raiz1 = complex(parte_real, parte_imaginaria)
        raiz2 = complex(parte_real, -parte_imaginaria)
        tipo_raices = "Raíces complejas"

    return raiz1, raiz2, tipo_raices, (x_vertice, y_vertice)

# Función para graficar la parábola
def graficar_parabola(a, b, c, raiz1, raiz2, vertice):
    x = np.linspace(vertice[0] - 10, vertice[0] + 10, 400)  # Rango en X centrado en el vértice
    y = a * x**2 + b * x + c  # Función cuadrática

    plt.figure(figsize=(8,6))  # Tamaño de la gráfica

    # Graficar la parábola
    plt.plot(x, y, label=f"$y = {a}x^2 + {b}x + {c}$", color="blue")

    # Graficar el vértice
    plt.scatter(vertice[0], vertice[1], color='red', zorder=3, label="Vértice")
    plt.text(vertice[0], vertice[1] + 0.5, f"V({vertice[0]:.2f}, {vertice[1]:.2f})",
             fontsize=10, ha='center', color='red', fontweight='bold')

    # Graficar las raíces si son reales
    if isinstance(raiz1, (int, float)) and isinstance(raiz2, (int, float)): 
        plt.scatter([raiz1, raiz2], [0, 0], color='green', zorder=3, label="Raíces")
        plt.text(raiz1, 0.5, f"R1({raiz1:.2f})", fontsize=10, ha='right', color='green', fontweight='bold')
        plt.text(raiz2, -0.5, f"R2({raiz2:.2f})", fontsize=10, ha='left', color='green', fontweight='bold')

    # Ejes y cuadrícula
    plt.axhline(0, color='black', linewidth=1)
    plt.axvline(0, color='black', linewidth=1)
    plt.grid(True, linestyle='--', alpha=0.6)

    # Etiquetas y título
    plt.xlabel("Eje X")
    plt.ylabel("Eje Y")
    plt.title("Gráfica de la Parábola")
    plt.legend()
    plt.show()
```

```python
a = float(input("Ingrese el coeficiente a: "))
b = float(input("Ingrese el coeficiente b: "))
c = float(input("Ingrese el coeficiente c: "))

# Verificar que 'a' no sea 0 (para evitar división por cero)
if a == 0:
    print("El coeficiente 'a' no puede ser 0 en una ecuación cuadrática.")
else:
    # Resolver la ecuación
    raiz1, raiz2, tipo_raices, vertice = resolver_ecuacion_cuadratica(a, b, c)

    # Mostrar resultados en consola
    print("\n--- Resultados ---")
    print(f"Tipo de raíces: {tipo_raices}")
    print(f"Raíz 1: {raiz1}")
    print(f"Raíz 2: {raiz2}")
    print(f"Vértice: ({vertice[0]:.2f}, {vertice[1]:.2f})")

    # Graficar la parábola
    graficar_parabola(a, b, c, raiz1, raiz2, vertice)
```
